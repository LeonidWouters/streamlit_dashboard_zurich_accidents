```python
zebrastreifen = gpd.read_file(r"C:\Users\leoni\Documents\HSLU\5. Semester\Informationsvisualisierung\Einzelprojekt\informationsvisualisierung-leonid-wouters\Daten\zebrastreifen.geojson")
```

```python
zebrastreifen.info()
```

```python
# nur id und geometry behalten
zebrastreifen = zebrastreifen[['id', 'geometry']]
```

```python
zebrastreifen
```

```python
# Erstelle eine Spalte "Zebrastreifen"
# Suche für jeden Unfall den nächsten Zebrastreifen und speichere die Distanz in die Spalte "Zebrastreifen"
unfaelle = unfaelle.to_crs(epsg=2056)
zebrastreifen = zebrastreifen.to_crs(epsg=2056)


unfaelle['Zebrastreifen'] = unfaelle.apply(lambda x: zebrastreifen.distance(x['geometry']).min(), axis=1)

unfaelle.head(2)
```

```python
unfaelle = unfaelle.to_crs(epsg=4326)
```

```python
# Unfaelle "Zebrastreifen" = True wenn Distanz kleiner als 5m
unfaelle['Zebrastreifen'] = unfaelle['Zebrastreifen'] < 2

unfaelle.head(2)
```

```python
unfaelle.head(2)
```

```python
# Plotte die Unfälle mit plotly
import plotly.express as px

fig = px.scatter_mapbox(unfaelle, lat=unfaelle.geometry.y, lon=unfaelle.geometry.x, color='Zebrastreifen', hover_name='Zebrastreifen')
fig.update_layout(mapbox_style="open-street-map")

fig.show()
```
