```python
# Lösche alle Spalten, die nicht benötigt werden
tempo = tempo.drop(columns=['ausnahmen_fahrverbot', 'fahrverbot_ssv', 'id','objectid', 'publiziert_vsi_datum', 'rechtskraeftig_datum', 'temporegime_technical', 'umgesetzt_datum'])
```

```python
tempo.head()
```

```python
# Lösche alle Einträge, die Fahrverbote enthalten
tempo = tempo[tempo['temporegime'] != 'Fahrverbot']
```

```python
# Füge dem unfaelle Datensatz eine Spalte 'Temporegime' hinzu
unfaelle['Temporegime'] = None
```

```python
from shapely.geometry import Point
# Wandle die Geometrie in 2D um, indem nur die x- und y-Koordinaten verwendet werden
unfaelle['geometry'] = unfaelle['geometry'].apply(lambda geom: Point(geom.x, geom.y))

# Setze die Geometrie auf den neuen 2D-Punkt
unfaelle = gpd.GeoDataFrame(unfaelle, geometry='geometry')
```

```python
unfaelle.head(2)
```

```python
tempo.head(2)
```

```python
unfaelle = unfaelle.set_crs(epsg=4326)
```

```python
import geopandas as gpd
from shapely.ops import nearest_points

# Stelle sicher, dass unfaelle und tempo dasselbe Koordinatensystem verwenden
unfaelle = unfaelle.to_crs(tempo.crs)

# Funktion zum Finden des nächstgelegenen Temporegimes für jeden Unfall
def find_nearest_temporegime(unfall_geom):
    nearest_line = tempo.geometry.unary_union
    nearest_point = nearest_points(unfall_geom, nearest_line)[1]
    
    # Suche nach der Linie, die dem nächstgelegenen Punkt entspricht
    nearest = tempo.distance(nearest_point).idxmin()
    return tempo.loc[nearest, 'temporegime']

# Füge das Temporegime zu den Unfällen hinzu
unfaelle['Temporegime'] = unfaelle.geometry.apply(find_nearest_temporegime)

# Ergebnisse überprüfen
unfaelle.head()
```
