```python
# Drop columns that are not needed
tempo_stadt = tempo_stadt.drop(columns=['rechtskraeftig_datum', 'umgesetzt_datum', 'publiziert_vsi_datum'])
```

```python
import folium
import geopandas as gpd

# Erstellen einer Folium-Karte
m = folium.Map(location=[47.3769, 8.5417], zoom_start=12)

# Funktion zur Farbabstufung basierend auf der Geschwindigkeit
def color_function(temporegime):
    # Dictionary zur Zuordnung von Geschwindigkeiten zu Farbwerten
    color_dict = {
        'T20': 'green',  # schnellste
        'T30': 'lightgreen',
        'T50': 'yellow',
        'T60': 'orange',
        'T80': 'lightred',
        'T100': 'red',  # langsamste
        'T120': 'darkred'
    }
    return color_dict.get(temporegime, 'gray')  # Standardfarbe für unbekannte Werte

# Hinzufügen der GeoJson-Daten für tempo_stadt
for _, row in tempo_stadt.iterrows():
    folium.GeoJson(
        row['geometry'],
        style_function=lambda feature, color=color_function(row['temporegime']): {
            'color': color,
            'weight': 2,
            'opacity': 0.7,
        },
        popup=folium.Popup(f"Geschwindigkeit: {row['temporegime']}", sticky=True)
    ).add_to(m)

# Karte anzeigen
m
```
