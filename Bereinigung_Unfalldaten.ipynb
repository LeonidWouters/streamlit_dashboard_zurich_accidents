```python
import geopandas as gpd
import pandas as pd
```

```python
unfaelle = gpd.read_file(r'C:\Users\leoni\Documents\HSLU\5. Semester\Informationsvisualisierung\Einzelprojekt\informationsvisualisierung-leonid-wouters\Daten\RoadTrafficAccidentLocations.json')
```

```python
# Behalte nur Daten aus dem Jahr 2015-2023 mit String
unfaelle = unfaelle[unfaelle['AccidentYear'].str.contains('2015|2016|2017|2018|2019|2020|2021|2022|2023')]
```

```python
# Behalte nur die Folgenden Spalten: 'AccidentUID','AccidentType_de', 'AccidentSeverityCategory_de', 'AccidentInvolvingPedestrian', 'AccidentInvolvingBicycle', 'AccidentInvolvingMotorcycle', 'RoadType_de', 'AccidentLocation_CHLV95_E', 'AccidentLocation_CHLV95_N', 'CantonCode', 'MunicipalityCode', 'AccidentYear', 'AccidentMonth_de', 'AccidentWeekDay_de', 'AccidentHour', 'AccidentHour_text', 'geometry'
unfaelle_bereinigt = unfaelle[['AccidentUID','AccidentType_de', 'AccidentSeverityCategory_de', 'AccidentInvolvingPedestrian', 'AccidentInvolvingBicycle', 'AccidentInvolvingMotorcycle', 'RoadType_de', 'AccidentLocation_CHLV95_E', 'AccidentLocation_CHLV95_N', 'CantonCode', 'MunicipalityCode', 'AccidentYear', 'AccidentMonth_de', 'AccidentWeekDay_de', 'AccidentHour', 'AccidentHour_text', 'geometry']]
```

```python
# Behalte nur Einträge für den Kanton Zürich
unfaelle_bereinigt = unfaelle_bereinigt[unfaelle_bereinigt['CantonCode'] == 'ZH']
```

```python
unfaelle_bereinigt.head()
```

```python
# Umbenennen der Spalten
unfaelle_bereinigt = unfaelle_bereinigt.rename(columns={'AccidentUID': 'UnfallID',
                                                        'AccidentType_de': 'Unfallart',
                                                        'AccidentSeverityCategory_de': 'Unfallschwere',
                                                        'AccidentInvolvingPedestrian': 'Fussgänger_beteiligt',
                                                        'AccidentInvolvingBicycle': 'Fahrrad_beteiligt',
                                                        'AccidentInvolvingMotorcycle': 'Motorrad_beteiligt',
                                                        'RoadType_de': 'Strassentyp',
                                                        'AccidentLocation_CHLV95_E': 'Unfallort_Lon',
                                                        'AccidentLocation_CHLV95_N': 'Unfallort_Lat',
                                                        'CantonCode': 'Kanton',
                                                        'MunicipalityCode': 'Gemeinde',
                                                        'AccidentYear': 'Jahr',
                                                        'AccidentMonth_de': 'Monat',
                                                        'AccidentWeekDay_de': 'Wochentag',
                                                        'AccidentHour': 'Stunde',
                                                        'AccidentHour_text': 'Stunde_text',})
```

```python
# Behalte nur Einträge mit der Gemeinde 0261
unfaelle_bereinigt = unfaelle_bereinigt[unfaelle_bereinigt['Gemeinde'] == '0261']
```
